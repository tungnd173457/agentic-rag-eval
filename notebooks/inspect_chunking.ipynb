# Inspect parent/child chunking

Load ONE gold-doc by path and print its parent/child chunks. Pure & free
(no Weaviate, no API). Set `DOC_PATH` then run the cells.

```python
import sys
from pathlib import Path

EVAL_DIR = Path.cwd().parent / "general-agent" / "eval"
sys.path.insert(0, str(EVAL_DIR))
import bootstrap  # noqa: F401 — sets sys.path so `ingestion` imports
from ingestion.splitters import split_document
```

```python
# Set the document to inspect (any .txt under gold-docs/).
DOC_PATH = Path.cwd().parent / "gold-docs" / "dsid_3c9fe2220c204f13b878bd7d3a168f91__2026-01-22-enterprise-prospect-procurement-via-aws-marketplace.txt"

markdown = DOC_PATH.read_text(encoding="utf-8", errors="replace")
parents, children = split_document(markdown)

print(f"file: {DOC_PATH.name}")
print(f"raw chars: {len(markdown)} | parents: {len(parents)} | children: {len(children)}\n")

for p in parents:
    print("=" * 88)
    print(f"PARENT p{p.position}  id={p.parent_id}  char_count={p.char_count}  children={len(p.child_ids)}")
    print("-" * 88)
    print(p.text)
    for c in children:
        if c.parent_id != p.parent_id:
            continue
        print(f"\n    · CHILD c{c.position}  id={c.child_id}  char_count={c.char_count}")
        print("      " + c.text.replace("\n", "\n      "))
    print()
```
